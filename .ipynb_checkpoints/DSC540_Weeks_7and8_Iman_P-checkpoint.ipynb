3. Discretization and Binning

```python
# checking on how many rows are available
import numpy as np
met_large_df = pd.read_csv(file_path, low_memory=False)
met_large_df.shape
```

```python
# Choosing only 10000 rows and selection of columns: Object Id, Title, object beginning data, and object end date to demonstrate binning
met_df = pd.read_csv(file_path, nrows=10000, low_memory=False)
# Choosing the selected columns for the task, and making sure there is no missing data
limited_df = met_df[['Object ID', 'Title', 'Object Begin Date', 'Object End Date']].dropna()
limited_df.shape
```

```python
# Defining bin edges for equal-width - 1000 years
min_year = int(limited_df['Object Begin Date'].min())
max_year = int(limited_df['Object End Date'].max())
bins = range(min_year, max_year + 1000, 1000)

# Creating labels
labels = [f"{i}-{i+999}" for i in bins[:-1]]

# Applying the cut function
limited_df['Date Range Group'] = pd.cut(
    limited_df['Object Begin Date'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

limited_df[['Object Begin Date', 'Date Range Group']].head()
```

```python
# Creating customized bins

# Defining customized bins
custom_bins = [-5000, 0, 1500, 1800, 1950, 2050]
age_labels = ['Ancient', 'Pre-Industrial', 'Industrial Era', 'Modern', 'Contemporary']

limited_df['Historical Era'] = pd.cut(
    limited_df['Object End Date'], 
    bins=custom_bins, 
    labels=age_labels,
    include_lowest=True
)

limited_df[['Object End Date', 'Historical Era']].head()
```

1. Create a hierarchical index  

```python
# Using Historical Era and Object ID to create a hierarchical index
hierarchical_df = limited_df.set_index(['Historical Era', 'Object ID'])

# Displaying the first five rows

hierarchical_df[['Title', 'Object End Date']].head(5)
```

```python
# Query data to see if there is data in the Ancient level in the hierarchical index
ancient_items = hierarchical_df.loc[['Ancient']].reset_index()
ancient_items[['Historical Era', 'Title', 'Object End Date']].head()
```

2. Combine and Merge Datasets (you will have to either create a new dataset from your existing data or create a relationship between the data I have provided)   


```python
# Creating Subset 1: Identifiers and Titles
df1 = limited_df[['Object ID', 'Title']].head(50)

# Creating Subset 2: Object Dates
df2 = limited_df[['Object ID', 'Object Begin Date', 'Object End Date']].head(50)

df1.head(5)
```

```python
df2.head(5)
```

```python
# Merging the two datasets back together on Object ID
merged_df = pd.merge(df1, df2, on='Object ID', how='inner')

print("Merged Dataset Shape")
print(merged_df.shape)
merged_df.head()
```

3. Reshape   


```python
# Using melt to reshape the data.
melted_df = pd.melt(
    limited_df.head(50), 
    id_vars=['Object ID', 'Title'], 
    value_vars=['Object Begin Date', 'Object End Date'],
    var_name='Date Type', 
    value_name='Year'
)

print("Melted/Reshaped Dataset")
melted_df.head()
```

4. Pivot the data

```python
print(limited_df.columns)
```

```python
# Creating a temporary subset to avoid index duplicates for pivoting
pivot_subset = limited_df.head(50).copy()

# Ensure Object ID is unique for the pivot
pivot_df = pivot_subset.pivot(
    index= 'Object ID', 
    columns='Historical Era', 
    values='Object End Date'
)

print("Pivoted Data Table Last rows out of 50")
pivot_df.tail()
```
